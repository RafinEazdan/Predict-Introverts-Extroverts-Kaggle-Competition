# Feature Engineering

```python
# # Create the interaction feature by multiplying existing columns
df_engg = df.copy()
df_engg_test = df2.copy()
df_engg['isolation_per_friend'] =  df_engg['Time_spent_Alone']/ df_engg['Friends_circle_size'].replace(0, np.nan)  # Avoid division by zero

df_engg_test['isolation_per_friend'] =  df_engg_test['Time_spent_Alone']/ df_engg_test['Friends_circle_size'].replace(0, np.nan)  # Avoid division by zero

df_engg_test.head()
```

```python
personality = df_engg['Personality']
df_engg = df_engg.drop(columns=['Personality', 'Time_spent_Alone', 'Friends_circle_size'])
df_engg_test = df_engg_test.drop(columns=['Time_spent_Alone', 'Friends_circle_size'])
df_engg['Personality'] = personality
df_engg.head()
```

# Converting to Numpy Array

```python
X = df.iloc[:,1:-1].values
y = df.iloc[:,-1].values
```

```python
XX = df2.iloc[:,1:].values
```

```python
print(X[:10])
print(y[:10])
```

```python
print(XX[:10])
```

# Imputation

```python
from sklearn.impute import SimpleImputer
# Use mean imputation for numeric columns only (columns 2, 5, 6)
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:, [2, 5, 6]])
X[:, [2, 5, 6]] = imputer.transform(X[:, [2, 5, 6]])
```

```python
# Use most_frequent imputation for categorical columns (columns 1, 4) and remaining numeric columns (0, 3)
imputer2 = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
X[:, [0, 1, 3, 4]] = imputer2.fit_transform(X[:, [0, 1, 3, 4]])
```

```python
# Apply same imputation to test data
# Mean imputation for numeric columns
imputer.fit(XX[:, [2, 5, 6]])
XX[:, [2, 5, 6]] = imputer.transform(XX[:, [2, 5, 6]])
# Most frequent imputation for categorical and remaining numeric columns  
XX[:, [0, 1, 3, 4]] = imputer2.fit_transform(XX[:, [0, 1, 3, 4]])
```

```python
print(X[:10])
print(y[:10])
```

# Encoding

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

encoding = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(), [1, 4])  # Apply to categorical columns only
    ],
    remainder='passthrough'  # Keep all other columns as they are
)
X = np.array(encoding.fit_transform(X))
```

```python
print(X[:10])
print(y[:10])
```

```python
XX = np.array(encoding.fit_transform(XX))
```

```python
print(XX[:10])
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
print(y[:10])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=0)
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_train_scaled[:, 4:] = scaler.fit_transform(X_train[:, 4:])
X_test_scaled = X_test.copy()
X_test_scaled[:, 4:] = scaler.transform(X_test[:, 4:])
X_scaled = X.copy()
X_scaled[:, 4:] = scaler.fit_transform(X[:, 4:])
```

```python
XX_scaled = XX.copy()
XX_scaled[:, 4:] = scaler.transform(XX[:, 4:])
```

```python
print(X_test_scaled[:10])
```

# Class Imbalance Check

```python
from collections import Counter
print(Counter(y))
```

```python
from imblearn.over_sampling import SMOTE
sm = SMOTE(random_state=0)
X_train_scaled_balanced, y_train_balanced = sm.fit_resample(X_train_scaled, y_train)
X_train_balanced, y_train_balanced = sm.fit_resample(X_train, y_train)
X_balanced, y_balanced = sm.fit_resample(X, y)
X_scaled_balanced, y_scaled_balanced = sm.fit_resample(X_scaled, y)
```

```python
from collections import Counter
print("After SMOTE:", Counter(y_train_balanced))
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state=0)
classifier.fit(X_train_scaled_balanced, y_train_balanced)
```

```python
y_pred = classifier.predict(X_test_scaled)
```

```python
y_pred_xx = classifier.predict(XX_scaled)
```

```python
print(y_pred_xx[:10])
test_predictions = le.inverse_transform(y_pred_xx)
print(test_predictions[:10])
```

# GridSearch Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# 1. Define your hyperparameter grid
param_grid = {
    'n_estimators': [100, 200, 300, 500],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

# 2. Create the classifier
rf = RandomForestClassifier(random_state=42)

# 3. GridSearchCV with cross-validation (5-fold)
grid_search = GridSearchCV(estimator=rf,
                           param_grid=param_grid,
                           cv=5,            # 5-fold cross-validation
                           n_jobs=-1,       # Use all cores
                           verbose=2,
                           scoring='f1')    # You can also use 'accuracy', 'roc_auc', etc.

# 4. Fit the grid search on full data
grid_search.fit(X_balanced, y_balanced)

# 5. Show best parameters
print("Best Parameters Found:")
print(grid_search.best_params_)

# 6. Train final model on full data using best params
best_rf = RandomForestClassifier(**grid_search.best_params_, random_state=42)
best_rf.fit(X_balanced, y_balanced)
```

```python
rfcv_y_pred_xx = best_rf.predict(XX)
print(rfcv_y_pred_xx[:10])
```

```python
rfcv_test_predictions = le.inverse_transform(rfcv_y_pred_xx)
print(rfcv_test_predictions[:10])
```

```python
# Assuming df2 has a column named "id" as its first column
ids = df2.iloc[:, 0]  # or df2['id'] if the column is named 'id'

# Create a new DataFrame for output
output_df = pd.DataFrame({
    'id': ids,
    'Personality': rfcv_test_predictions
})
# Save to CSV
output_df.to_csv('rfcv_submission.csv', index=False)

# Optional: preview the first few rows
print(output_df.head())
```
